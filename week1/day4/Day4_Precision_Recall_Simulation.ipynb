# Interactive Information Retrieval Metrics Calculator

This interactive notebook demonstrates how **Precision**, **Recall**, and the **F1-Score** change dynamically based on the outcomes of a retrieval system (like a search engine or a RAG vector database).

Run the code cell below to load the interactive sliders.

```python
import ipywidgets as widgets
from IPython.display import display, HTML

def calculate_metrics(tp, fp, fn):
    # Handle divide-by-zero gracefully
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    # Format the output as HTML for better visualization
    html_out = f"""
    <html>
    <div style='background-color: #f8f9fa; padding: 20px; border-radius: 8px; border: 1px solid #dee2e6; margin-top: 20px; width: 500px;'>
        <h3 style='margin-top:0; color: #212529;'>Calculated Metrics</h3>
        <ul style='font-size: 16px; line-height: 1.8; color: #495057;'>
            <li><b>Precision</b> $\left(\frac{tp}{tp+fp}\right)$: <span style='color: #0d6efd; font-size: 18px;'><b>{precision:.2%}</b></span></li>
            <li><b>Recall</b> $\left(\frac{tp}{tp+fn}\right)$: <span style='color: #198754; font-size: 18px;'><b>{recall:.2%}</b></span></li>
            <li><b>F1-Score</b> (Harmonic Mean): <span style='color: #6f42c1; font-size: 18px;'><b>{f1:.2%}</b></span></li>
        </ul>
    </div>
    </html>
    """
    display(HTML(html_out))

# Create interactive sliders
style = {'description_width': 'initial'}
layout = widgets.Layout(width='600px')

tp_slider = widgets.IntSlider(value=5, min=0, max=100, description='True Positives (Retrieved & Relevant):', style=style, layout=layout)
fp_slider = widgets.IntSlider(value=3, min=0, max=100, description='False Positives (Retrieved & Irrelevant):', style=style, layout=layout)
fn_slider = widgets.IntSlider(value=5, min=0, max=100, description='False Negatives (Missed & Relevant):', style=style, layout=layout)

# Bind the sliders to the calculation function
out = widgets.interactive_output(calculate_metrics,
                                 {'tp': tp_slider, 'fp': fp_slider, 'fn': fn_slider}
                                 )

# Display the UI
display(widgets.VBox([tp_slider, fp_slider, fn_slider, out]))
```
