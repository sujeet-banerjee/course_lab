# Autoregressive LLM Token Generation Simulator

This interactive notebook demonstrates how Large Language Models (like GPT, Claude, or LLaMA) generate text one token at a time and feed it back into their context window in a continuous loop.

### Instructions:
1. Run the code cell below.
2. Click **Generate Next Token** to watch the LLM predict the next word and append it to the growing input sequence.

```python
import ipywidgets as widgets
from IPython.display import display, HTML
import time

# --- Data State ---
initial_context = ['The', 'robot', 'learned', 'to']
target_tokens = ['code', 'in', 'Python', '&lt;STOP&gt;']
current_context = initial_context.copy()
step = 0

# --- CSS Styling ---
style = """
<style>
    .token { padding: 6px 12px; margin: 4px; border-radius: 6px; background: #e3f2fd; display: inline-block; font-family: monospace; border: 1px solid #90caf9; font-size: 16px; }
    .new-token { background: #c8e6c9; border-color: #4caf50; font-weight: bold; transition: background 0.5s ease; }
    .container { padding: 15px; border: 2px dashed #ccc; border-radius: 8px; margin-bottom: 15px; }
</style>
"""

# --- UI Components ---
context_html = widgets.HTML(value="")
llm_html = widgets.HTML(value="<div class='container'><i>Waiting for user to trigger generation...</i></div>")

btn_gen = widgets.Button(description="Step: Generate Next Token", button_style='primary', layout=widgets.Layout(width='auto'))
btn_reset = widgets.Button(description="Reset Simulation", button_style='warning', layout=widgets.Layout(width='auto'))

def render_context(context_list, highlight_last=False):
    html = style + "<div class='container'><h3>Input Sequence (Context Window)</h3><div>"
    for i, token in enumerate(context_list):
        css_class = "new-token" if highlight_last and i == len(context_list) - 1 else "token"
        html += f"<span class='{css_class}'>{token}</span>"
    html += "</div></div>"
    return html

def update_ui(highlight=False):
    context_html.value = render_context(current_context, highlight)

def on_generate(b):
    global step, current_context
    if step < len(target_tokens):
        next_token = target_tokens[step]
        
        # 1. Simulate prediction/processing
        llm_html.value = "<div class='container'><h3>LLM Processing...</h3><p>Calculating self-attention and probability distributions...</p></div>"
        time.sleep(0.6) # small delay for visual effect
        
        # 2. Show Output Token
        llm_html.value = f"<div class='container'><h3>Output Token</h3><span class='token new-token'>{next_token}</span><p><i>Appending to input context for next iteration...</i></p></div>"
        time.sleep(0.4)
        
        # 3. Append and update context window
        current_context.append(next_token)
        step += 1
        update_ui(highlight=True)
        
        if next_token == '&lt;STOP&gt;':
            btn_gen.disabled = True
            llm_html.value = "<div class='container'><h3>Generation Halted</h3><span class='token new-token'>&lt;STOP&gt;</span><p><b>Model predicted the STOP token. Loop finished!</b></p></div>"

def on_reset(b):
    global step, current_context
    current_context = initial_context.copy()
    step = 0
    btn_gen.disabled = False
    llm_html.value = "<div class='container'><i>Waiting for user to trigger generation...</i></div>"
    update_ui()

# --- Bind Events & Display ---
btn_gen.on_click(on_generate)
btn_reset.on_click(on_reset)

update_ui()
display(widgets.VBox([
    context_html, 
    llm_html, 
    widgets.HBox([btn_gen, btn_reset])
]))
```
